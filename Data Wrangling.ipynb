```python
categorical_variables=['County', 'City', 'State', 'Postal Code', 
                       'Model Year', 'Make','Model', 'Electric Vehicle Type', 'Clean Alternative Fuel Vehicle (CAFV) Eligibility',
                      'Legislative District', 'Electric Utility','VIN (1-10)']


numerical_variables=['Electric Range', 'Base MSRP', '2020 Census Tract']
```

```python
plt.figure(figsize=(20, 10))
ind=1
for  variable in (numerical_variables):
    plt.subplot(2, 3, ind)
    sns.boxplot(x=df[variable],color="blue")
    plt.title(variable)
    ind+=1
    plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
for variable in categorical_variables:
    print(df[variable].value_counts())
    print("-"*50)
```
